```python
!pip install -q SpeechRecognition

import speech_recognition as sr
from google.colab import files
import os

# Step 1: Open an upload dialog to choose and upload a file from your PC
print("Please choose an audio file from your computer...")
uploaded = files.upload()

# Get the filename of the uploaded file automatically
if uploaded:
    audio_file = list(uploaded.keys())[0]
    print(f"\nSuccessfully uploaded: {audio_file}")

    # Create recognizer
    recognizer = sr.Recognizer()

    try:
        # Read the audio file
        with sr.AudioFile(audio_file) as source:
            print("Reading audio file...")
            audio = recognizer.record(source)

        # Convert speech to text
        print("Converting speech to text...")
        text = recognizer.recognize_google(
            audio,
            language="en-IN"
        )

        # Display the result
        print("\nSpeech to Text Output:")
        print(text)

        # Save text to a file
        with open("output.txt", "w", encoding="utf-8") as f:
            f.write(text)

        print("\nText saved successfully to output.txt")

    except sr.UnknownValueError:
        print("Speech could not be understood.")

    except sr.RequestError as e:
        print("Speech service error. Check your internet connection.")
        print("Details:", e)

    except (ValueError, OSError) as e:
        print("Audio file error (ensure it is a valid WAV format):", e)

    except Exception as e:
        print("Unexpected error:", e)
else:
    print("No file was uploaded.")
```
